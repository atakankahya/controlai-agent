### Step 2: Initialize the ControlAI Agent

```python
import sys
from pathlib import Path
from IPython.display import display, Markdown, Image
import matplotlib.pyplot as plt

from controlai_agent.orchestrator import ControlAIAgent

print("Loading ControlAI Engine with Qwen 3B...")
agent = ControlAIAgent(model_path="Qwen/Qwen2.5-3B-Instruct")
print("ControlAI Engine ready!")
```

### Step 3: Ask Any Control Engineering Question
You can edit the prompt below to test your own dynamical systems, transfer functions, or optimal control problems.

```python
# Enter your engineering query here:
prompt = "Design an LQR controller for A=[[0, 1], [-2, -3]], B=[[0], [1]], Q=diag([10, 1]), R=1 and simulate the step response."

print(f"User Query: {prompt}\n")
result = agent.run(prompt)

# Display formatted response
display(Markdown(result.final_response))

# Display any generated simulation plots
for plot_path_rel in result.plots:
    clean_name = Path(plot_path_rel).name
    local_path = Path("outputs/plots") / clean_name
    if local_path.exists():
        print(f"\nGenerated Simulation Plot: {clean_name}")
        display(Image(filename=str(local_path)))
```
